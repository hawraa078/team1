```python
df_sorted=df.sort_values(by='daily_steps')
plt.bar(df_sorted['daily_steps'],df_sorted['calories_burned'])
plt.xlabel('daily_steps')
plt.ylabel('calories_burned')
plt.show()
```

```python
mean_of_sleep=df.groupby('age')['hours_sleep'].mean()
plt.plot(mean_of_sleep.index,mean_of_sleep.values)
plt.xlabel('age')
plt.ylabel('hours_sleep')
plt.show()
```

```python
df.groupby("stress_level")["hours_sleep"].mean().plot(kind="bar")

plt.title("Average Sleep Hours by Stress Level")
plt.xlabel("Stress Level")
plt.ylabel("Average Sleep Hours")
plt.show()
```

```python
df.groupby("activity_type")["calories_burned"].mean().sort_values().plot(kind="barh", figsize=(8,5))

plt.title("Average Calories Burned by Activity")
plt.xlabel("Calories")
plt.show()
```
